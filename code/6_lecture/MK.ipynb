```python
import random
import math
import numpy as np
import matplotlib.pyplot as plt

def func(x):
    return np.sin(x)/(1+np.cos(x)**2)

def monte_carlo(num_points, boundary):
    """
    Вычисляет площадь круга методом Монте-Карло
    """
    left, right = boundary
    points_inside = 0
    for _ in range(num_points):

        x = random.uniform(left, right)
        y = random.uniform(0, 1)
        
        # Проверяем, попадает в нужную область
        if y < func(x):
            points_inside += 1
            x_array.append(x)
            y_array.append(y)
            
    # Вычисляем площадь квадрата и оцениваем площадь под кривой
    square_area = (right - left) * 1 
    estimated_area = (points_inside / num_points) * square_area
    
    return estimated_area

# Пример использования

x_array = []
y_array = []
boundary = (0, np.pi)
error_array = []
num_points_array = [10, 100, 1000, 10000, 100000, 1000000]

for _ in num_points_array:
    estimated = monte_carlo(_, boundary = boundary)
    error = abs(estimated - np.pi/2)  # Предположим, что точная площадь равна 1.0
    error_array.append(error)

plt.loglog(num_points_array, error_array)
plt.xlabel("Number of Points")
plt.ylabel("Error")
plt.title("Convergence of Monte Carlo Method")
print(f"Оценка площади: {estimated:.6f}")
plt.show()
plt.scatter(x_array, y_array)
```
